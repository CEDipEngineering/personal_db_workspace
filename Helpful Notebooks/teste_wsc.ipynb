```python
from databricks.sdk import WorkspaceClient

# Initialize with your Databricks host and token or rely on default authentication
w = WorkspaceClient()

from databricks.sdk.service import sql
import pandas as pd

# Replace with your actual SQL statement and warehouse ID
sql_statement = "SELECT * FROM cedip.default.iris_data LIMIT 10"
warehouse_id = "8baced1ff014912d"

# Execute the statement
response = w.statement_execution.execute_statement(
    statement=sql_statement,
    warehouse_id=warehouse_id,
    # Optional: Specify catalog and schema
    catalog="cedip",
    schema="default"
)

# Access results if available synchronously
if response.result:
    columns = [col.name for col in response.result.columns]
    data = response.result.data_array
    df = pd.DataFrame(data, columns=columns)
    display(df)
else:
    print(f"Statement ID for asynchronous execution: {response.statement_id}")
    # You would then use get_statement to poll for results
```
